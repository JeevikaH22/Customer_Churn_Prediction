### Formatting-

```python
df.head()
```

```python
cat_col=["MultipleLines","InternetService","Contract","PaymentMethod","OnlineSecurity"]

for col in range(len(cat_col)):
    print(f"{col} = {df[cat_col[col]].unique()}")
```

```python
df.replace({"No phone service":"No", "No internet service": "No"}, inplace=True)
```

```python
df['PaymentMethod'] = df['PaymentMethod'].str.replace(" (automatic)", "", regex=False).str.strip()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"].str.strip(), errors="coerce")
```

```python
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df.to_csv('../data/Telco-Customer-Churn-cleaned.csv', index=False)
```
